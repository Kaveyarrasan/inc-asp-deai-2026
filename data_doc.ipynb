## KaTeX table examples

**Important:** KaTeX does **not** support `\begin{table}` or `\begin{tabular}` — those are LaTeX text/float environments, not math. For table-like layouts use the `array` environment with `|` column separators and `\hline`, or use a matrix environment.

### 1. Array with full borders (closest to a table)

$$
\begin{array}{|c|c|c|}
\hline
\text{Metric} & \text{Value} & \text{Unit} \\
\hline
\alpha & 0.05 & - \\
\beta & 1.96 & - \\
\text{count} & 42 & \text{rows} \\
\hline
\end{array}
$$

### 2. Array with partial borders and mixed column alignment

Column spec `r|l` = right-aligned | vertical bar | left-aligned. Other options: `l`, `c`, `r`, `|`, `||`.

$$
\begin{array}{r|l}
\hline
\text{Item} & \text{Value} \\
\hline
\alpha & 0.05 \\
\beta & 1.96 \\
\hline
\end{array}
$$

### 3. Matrix environments (grid layouts)

$$
\begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}
\quad
\begin{bmatrix} a & b \\ c & d \end{bmatrix}
\quad
\begin{vmatrix} 1 & 2 \\ 3 & 4 \end{vmatrix}
$$

### 4. Inline table-like array

$\left(\begin{smallmatrix} a & b \\ c & d \end{smallmatrix}\right)$ small matrix fits inline.

### Reference

Full list of supported environments (`array`, `matrix`, `pmatrix`, `bmatrix`, `cases`, `aligned`, `gather`, ...) and functions: [KaTeX supported functions](https://katex.org/docs/supported)
