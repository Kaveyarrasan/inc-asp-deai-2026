8. dbutils widget Usecase – Move the above processed File to Archive

```python
folder_stage = ['Raw', 'Processed', 'Archive']
folder_stage_path = [raw_path, processed_path, archive_path]

db_raw_path = dbutils.fs.ls(raw_path)[0].path
db_processed_path = dbutils.fs.ls(processed_path)[0].path
# db_archive_path = dbutils.fs.ls(archive_path)[0].path
print(db_raw_path)
print(db_processed_path)
# print(db_archive_path)

# for path in folder_stage_path:
#   # print(f"Path: {dbutils.fs.ls(path)}")
#   if len(dbutils.fs.ls(path)) > 0:
#     # print(f"Path: {dbutils.fs.ls(path)[0].path}")
#     folder_stage_path.append(dbutils.fs.ls(path)[0].path)
#   # else:
#   #   print("Path is empty")

# for i in dbutils.fs.ls(processed_path):
#   print(f"Path: {i}")

zip_stages = zip(folder_stage, directory_list)

for stage, path in zip_stages:
  print(f"Stage: {stage}, Path: {path}")
```

```python
dbutils.widgets.dropdown('Stage', 'Raw', folder_stage)
widg_stage = dbutils.widgets.get('Stage')
print(f"Stage: {widg_stage}")

if widg_stage == 'Raw':
    print("File is in raw state")
elif widg_stage == 'Processed':
    dbutils.fs.mv(db_processed_path, archive_path, recurse=True)
    print("File is processed and moved to archive")
elif widg_stage == 'Archive':
    print("File is archived")
```
