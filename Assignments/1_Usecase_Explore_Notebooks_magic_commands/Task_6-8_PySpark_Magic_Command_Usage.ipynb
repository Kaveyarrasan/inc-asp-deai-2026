```python
%python

file_location = "dbfs:/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"
df1 = spark.read.format("csv").option("Header", True).option("InferSchema", True).load(file_location)
# df1 = spark.read.format("csv").option("Header", True).load(file_location)

display(df1)
```
